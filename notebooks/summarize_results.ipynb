## LPB and UPB Combined AutoIF Results

Combine the LPB and UPB results for the AutoIF dataset into a unified view. This cell loads both datasets, concatenates them, and plots a grouped boxplot to contrast the two bound types.

```python
import os
import pandas as pd
from src.safety_evaluation.utils.utils import get_merged_calibration_result_path, get_merged_upb_calibration_result_path
from common_eval_utils import load_coverage_data, generate_coverage_boxplots, COLOR_MAP

# 1. Configuration
figures_dir = "./figures/summarized_lpb_upb_calibration_results/main"
is_main = True

# 2. Maps for LPB
lpb_autoif_map = {
    'dataset_autoif_attack_autoif_helper_qwen25_14b_instruct_lm_target_qwen25_14b_instruct_judge_autoif_20.0_3000_0.56_10.0': "Qwen2.5",
}
lpb_display_name_map = {
    'uncalibrated': 'Uncalibrated',
    'calibration_optimized_allocation': 'Optimized',
    'calibration_projected_optimization_platt_prob_n1_100_allocation': 'Ours',
}

# 3. Maps for UPB
upb_autoif_map = {
    'dataset_autoif_attack_autoif_helper_qwen25_14b_instruct_lm_target_qwen25_14b_instruct_judge_autoif_20.0_3000_0.97_10.0': "Qwen2.5",
}
upb_display_name_map = {
    'uncalibrated': 'Uncalibrated',
    'calibration_optimized_allocation': 'Optimized',
    'calibration_projected_optimization_platt_prob_allocation': 'Ours',
}

print("--- Generating Combined LPB and UPB Boxplots ---")

# 4. Extract and align data
df_autoif_lpb_comb = load_coverage_data(lpb_autoif_map, lpb_display_name_map, "autoif", 
                                        get_merged_calibration_result_path, target_coverages=[90.0], bound_type='LPB')
df_autoif_upb_comb = load_coverage_data(upb_autoif_map, upb_display_name_map, "autoif", 
                                        get_merged_upb_calibration_result_path, target_coverages=[70.0], bound_type='UPB')

dfs_to_concat = []
if df_autoif_lpb_comb is not None: dfs_to_concat.append(df_autoif_lpb_comb)
if df_autoif_upb_comb is not None: dfs_to_concat.append(df_autoif_upb_comb)

# 5. Execution 
if dfs_to_concat:
    df_autoif_combined = pd.concat(dfs_to_concat, ignore_index=True)
    generate_coverage_boxplots(df_autoif_combined, figures_dir, "autoif", COLOR_MAP, 
                               target_budget=20, target_coverage=None, is_main=is_main, x_col='Bound Type')
else:
    print("No LPB or UPB data found to combine.")
```
