## Coverage bound results
In this notebook, we visualize the coverage bound guaranteed in our theory and compare it to the coverage bound from prior works.

## Imports

Import standard numerical libraries and matplotlib for plotting.

```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

## Define Parameters

Define constants for the theoretical bounds: calibration set size ($n$), target error rate ($\alpha$), confidence level ($\delta$), and the range for the maximum weight ($\gamma$).

```python
n = 3000                 # Calibration set size (|I|)
alpha = 0.1              # Target error rate (1 - alpha = 0.9)
delta = 0.05             # Confidence level
gamma = np.linspace(1, 20, 500)  # Maximum weight (w_max), ranging from 1 to 20

# Mean weight assumption: Uniform distribution from 1 to gamma
w_mean = (1 + gamma) / 2
```

## Compute Lower Bounds

Calculate the coverage gaps and lower bound percentages for both our proposed approach (using mean weight) and the previous baseline (using maximum weight).

```python
log_delta_inv = np.log(1 / delta)

# "Ours" Coverage Gap and Lower Bound
term1_ours = log_delta_inv / (3 * n)
term2_ours = np.sqrt((log_delta_inv**2) / (9 * n**2) + (2 * (w_mean - alpha**2) * log_delta_inv) / n)
gap_ours = term1_ours + term2_ours
bound_ours = (1 - alpha - gap_ours) * 100  # Convert to percentage

# "Previous" Coverage Gap and Lower Bound
gap_prev = np.sqrt(((2 * gamma**2 + 5) / n) * log_delta_inv)
bound_prev = (1 - alpha - gap_prev) * 100  # Convert to percentage
```

## Plot the Figure

Generate a line plot comparing the theoretical coverage bound of our method versus the prior method across different values of $\gamma$ (max weight).

```python
plt.figure(figsize=(8, 4))

# Plot the bounds
plt.plot(gamma, bound_ours, label='Proposed Bound (Mean Weight)', color='tab:green', linewidth=2.5)
plt.plot(gamma, bound_prev, label='Prior Bound (Max Weight)', color='tab:blue', linewidth=2.5, linestyle='--')

# Add nominal coverage line
plt.axhline(y=(1 - alpha) * 100, color='black', linestyle=':', linewidth=2, label='Nominal Coverage (90%)')

# Formatting the plot for academic presentation
plt.xlabel(r'$w_{\max}$ ($\gamma$)', fontsize=22)
plt.ylabel('Coverage Bound (%)', fontsize=22)
# plt.title('Coverage Bound vs. Max. Weight', fontsize=24)
plt.xticks(fontsize=20)
plt.yticks(fontsize=20)

# Set axes limits for better visibility
plt.xlim([1, 20])
plt.ylim([min(np.min(bound_prev), 75), 95])  # Adjust lower y-limit dynamically, cap top at 95%

plt.legend(fontsize=18, loc='lower left')
plt.grid(True, alpha=0.3)
plt.tight_layout()
os.makedirs("figures", exists_ok=True)
plt.savefig("figures/coverage_bounds.png", dpi=300, bbox_inches='tight')
plt.show()
```
